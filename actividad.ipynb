### **Actividad 2:** 
Acceso y selección de elementos.
* Usa una Series con tareas (que haceres cotidianos) de una semana
* Accede a diferentes elementos a elección.
* Cambia los valores de los dias pares.


```python
import pandas as pd

tareas_semana = pd.Series([
    "Lunes: Hacer ejercicio",
    "Martes: Ir al supermercado",
    "Miércoles: Estudiar",
    "Jueves: Ir a taekwondo",
    "Viernes: Trabajar en el proyecto",
    "Sábado: Limpiar la casa",
    "Domingo: Descansar"
])

lunes_tarea = tareas_semana[0]
miercoles_tarea = tareas_semana[2]

tareas_semana[1] = "Martes: Ir a pagar los impuestos "
tareas_semana[3] = "Jueves: Ir a zumba"
tareas_semana[5] = "Sábado: Ir a la casa de mi tía"

print(lunes_tarea)
print(miercoles_tarea)
print(tareas_semana)
```

### **Actividad 3:**
Operaciones aritméticas. 
* Realiza las operaciones basicas con 2 series cuales quieran.
* Eleva los valores de la primer serie a 3 potencias difetentes.

```python
serie1 = pd.Series([2,3])
serie2 = pd.Series([3,4])

suma = serie1 + serie2
resta = serie1 - serie2
multiplicacion = serie1 * serie2
division = serie1 / serie2

print("Suma: ", suma)
print("Resta: ", resta)
print("Multiplicación: ", multiplicacion)
print("División: ", division)

potencia2 = serie1 ** 2  
potencia3 = serie1 ** 3  
potencia4 = serie1 ** 4 


print("Potencia 2: ", potencia2)
print("Potencia 3: ", potencia3)
print("Potencia 4: ", potencia4)
```

### **Actividad 4:** 
Filtrado con condiciones.
* Crea una Series de edades de 10 personas.
* Muestra solo a los mayores de 18.
* Reemplaza a los menores por "Menor de edad".

```python
import pandas as pd

edades = pd.Series([12, 25, 17, 30, 45, 16, 19, 21, 13, 29])

mayores_18 = edades[edades > 18]
print("Mayores de 18 años:")
print(mayores_18)

menores_de_edad_reemplzados = edades.apply(lambda x: x if x > 18 else "Menor de edad")
print("Serie con menores reemplazados:")
print(menores_de_edad_reemplzados)
```

### Actividad 5:
Estadísticas básicas.

* Crea una Series con las notas de 10 estudiantes.
* Calcula la media, mediana, desviación estándar y máximo.
* Muestra cuántos estudiantes aprobaron si el mínimo para aprobar es 6.
* Ordenar las notas de manera ascendente y descendente. 

```python
notas = pd.Series([7, 5, 8, 6, 9, 4, 10, 6, 5, 7])

media = notas.mean()
mediana = notas.median()
desviacion_estandar = notas.std()
maximo = notas.max()

aprobados = notas[notas >= 6].count()

print ("media:", media)
print ("mediana:", mediana)
print ("desviación estandar:", desviacion_estandar)
print ("maximo:", maximo)
print ("aprobados:", aprobados)
```
